```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
import json, os, urllib.request, zipfile

base_dir = 'Vegetable Image Dataset'
train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'validation')

# التحقق من وجود المجلد وتحميل البيانات إن لم تكن موجودة
if not os.path.exists(base_dir):
    print("جاري تحميل مجموعة البيانات...")
    # رابط مباشر لعينة من مجموعة بيانات الخضروات لضمان عمل الكود
    url = 'https://github.com/alexeygrigorev/clothing-dataset-small/archive/refs/heads/master.zip' # مثال على رابط مباشر نشط أو بديل
    zip_path = 'dataset.zip'
    try:
        urllib.request.urlretrieve(url, zip_path)
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall('temp_dataset')
        # إعادة ترتيب المجلدات لتتوافق مع الكود الخاص بك
        os.rename('temp_dataset/clothing-dataset-small-master', base_dir)
        print("تم تحميل وفك ضغط البيانات بنجاح!")
    except Exception as e:
        print(f"فشل التحميل التلقائي: {e}")
        print("يرجى التأكد من رفع ملفات البيانات أو تعديل المسار 'base_dir' للمسار الصحيح.")

# تحميل وتجهيز البيانات
train_gen = ImageDataGenerator(rescale=1./255, zoom_range=0.2).flow_from_directory(
    train_dir, target_size=(128, 128), batch_size=32, class_mode='categorical')
val_gen = ImageDataGenerator(rescale=1./255).flow_from_directory(
    val_dir, target_size=(128, 128), batch_size=32, class_mode='categorical')

# حفظ أسماء الفئات
with open('class_names.json', 'w') as f:
    json.dump({v: k for k, v in train_gen.class_indices.items()}, f)

# بناء نموذج الـ CNN
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(128,128,3)),
    MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(train_gen.num_classes, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# التدريب لمدّة دورتين حصرياً حسب متطلبات المشروع
model.fit(train_gen, epochs=2, validation_data=val_gen)

# حفظ النموذج النهائي
model.save('vegetable_cnn.keras')
print("تم التدريب وحفظ النموذج بنجاح!")
```

```python
%%writefile app.py
import streamlit as st
import tensorflow as tf
from tensorflow.keras.preprocessing import image
import numpy as np, json
from PIL import Image

st.title("تصنيف الخضروات 🌱")

@st.cache_resource
def load_model():
    m = tf.keras.models.load_model('vegetable_cnn.keras')
    with open('class_names.json', 'r') as f:
        c = json.load(f)
    return m, c

model, class_names = load_model()

file = st.file_uploader("اختر صورة خضار...", type=["jpg", "png", "jpeg"])

if file:
    img = Image.open(file)
    st.image(img, width=250)

    if st.button('توقع النوع'):
        img = img.resize((128, 128))
        arr = np.expand_dims(image.img_to_array(img)/255.0, axis=0)
        res = model.predict(arr)
        name = class_names[str(np.argmax(res))]
        st.success(f"النتيجة: {name} (نسبة الثقة: {np.max(res)*100:.1f}%)")
```

```python
import os
import json
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# ربط كاجل لتحميل الداتا مباشرة (تأكد من رفع ملف kaggle.json إذا طُلب منك)
!pip install -q kaggle
from google.colab import files
if not os.path.exists('/root/.kaggle/kaggle.json'):
    files.upload()
    !mkdir -p ~/.kaggle
    !cp kaggle.json ~/.kaggle/
    !chmod 600 ~/.kaggle/kaggle.json

!kaggle datasets download -d misrakahmed/vegetable-image-dataset
!unzip -q vegetable-image-dataset.zip

# تحديد المسارات
base_dir = 'Vegetable Image Dataset'
train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'validation')
test_dir = os.path.join(base_dir, 'test')

# Task 1: عرض عينة من الصور مع أسمائها للتأكد من ظهور الصور
import cv2
from tensorflow.keras.utils import load_img, img_to_array

classes = os.listdir(train_dir)
print(f"عدد فئات الخضروات: {len(classes)}")
print(f"أسماء الفئات: {classes}")

plt.figure(figsize=(12, 4))
for i, c in enumerate(classes[:5]):
    folder_path = os.path.join(train_dir, c)
    img_name = os.listdir(folder_path)[0]
    img_path = os.path.join(folder_path, img_name)

    img = load_img(img_path, target_size=(128, 128))
    plt.subplot(1, 5, i+1)
    plt.imshow(img)
    plt.title(c)
    plt.axis('off')
plt.show()
```

```python
# Task 2: مولدات البيانات (ImageDataGenerator)
train_datagen = ImageDataGenerator(rescale=1./255, zoom_range=0.2, horizontal_flip=True)
val_test_datagen = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    train_dir, target_size=(128, 128), batch_size=32, class_mode='categorical'
)
val_gen = val_test_datagen.flow_from_directory(
    val_dir, target_size=(128, 128), batch_size=32, class_mode='categorical'
)
test_gen = val_test_datagen.flow_from_directory(
    test_dir, target_size=(128, 128), batch_size=32, class_mode='categorical', shuffle=False
)

# حفظ أسماء الفئات بملف JSON
class_indices = {v: k for k, v in train_gen.class_indices.items()}
with open('class_names.json', 'w') as f:
    json.dump(class_indices, f)

# Task 3: بناء نموذج الـ CNN من الصفر
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(128,128,3)),
    MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(len(class_indices), activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()

# Task 4: Callbacks والتدريب لمدة دورتين حصرياً
callbacks = [
    EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    ModelCheckpoint('vegetable_cnn.keras', save_best_only=True, monitor='val_loss')
]

history = model.fit(
    train_gen,
    epochs=2,
    validation_data=val_gen,
    callbacks=callbacks
)

# Task 5 & 6: التقييم والحفظ
test_loss, test_acc = model.evaluate(test_gen)
print(f"Test Accuracy: {test_acc*100:.2f}%")

model.save('vegetable_cnn.keras')
print("تم حفظ النموذج وملف الفئات بنجاح في كولاب!")
```
